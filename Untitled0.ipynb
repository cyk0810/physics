```python
import math

# 상수 선언
PI = 3.14159265
g = -9.8  # 중력가속도 (m/s^2)

# 1. 각도 입력
theta_deg = float(input("각도를 입력하세요 : "))

# 2. 초기속도 입력
v0 = float(input("초기속도를 입력하세요 : "))

# 3. 각도 radian으로 변환
theta_rad = theta_deg * PI / 180.0

# 4. 초기 속도의 x, y 성분
v0x = v0 * math.cos(theta_rad)
v0y = v0 * math.sin(theta_rad)

# 5. 최고점 높이
h_peak = (v0y ** 2) / (2 * -g)

# 6. 최고점에 도달하는 시간
t_peak = v0y / -g

# 7. 수평 도달 거리
t_total = 2 * t_peak
x_max = v0x * t_total

# 8. 지면 도달 시 속도 성분
vx_final = v0x
vy_final = v0y
v_final = math.sqrt(vx_final ** 2 + vy_final ** 2)

# 9. 결과값 출력
print(f"1) 초기속도 성분: v0x = {v0x:.2f} m/s, v0y = {v0y:.2f} m/s")
print(f"2) 최고점 높이: {h_peak:.2f} m")
print(f"3) 최고점 도달 시간: {t_peak:.2f} s")
print(f"4) 수평 도달 거리: {x_max:.2f} m")
print(f"5) 지면 도달 시 속도 성분: Vx = {vx_final:.2f} m/s, Vy = {vy_final:.2f} m/s")
print(f"   속도 벡터 크기 V = {v_final:.2f} m/s")
```
