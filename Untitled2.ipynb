```python
import math

def main():
    # 0 초기 조건 입력
    v0 = float(input("초기 속도를 입력하세요 (m/s): "))
    angle_deg = float(input("발사 각도를 입력하세요 (도): "))
    y0 = float(input("초기 높이를 입력하세요 (m): "))
    g = 9.81  # 중력 가속도 (m/s^2)

    # 1 각도를 라디안으로 변환
    angle_rad = math.radians(angle_deg)

    # 3 초기 속도의 x, y 성분 계산
    vx = v0 * math.cos(angle_rad)
    vy = v0 * math.sin(angle_rad)

    # 4 지면에 도달할 때까지 걸리는 시간 계산
    # y = y0 + vy*t - 0.5*g*t^2 = 0
    a = -0.5 * g
    b = vy
    c = y0

    discriminant = b ** 2 - 4 * a * c

    if discriminant < 0:
        print("해가 존재하지 않습니다.")
        return

    t1 = (-b + math.sqrt(discriminant)) / (2 * a)
    t2 = (-b - math.sqrt(discriminant)) / (2 * a)
    time = t1 if t1 > 0 else t2

    print(f"(a) 지면에 도달할 때까지 걸리는 시간: {time:.2f} 초")

    # 5 수평 도달 거리 계산
    distance = vx * time
    print(f"(b) 수평 도달 거리: {distance:.2f} 미터")

    # 6 충돌 순간 속도 크기 계산
    vy_final = vy - g * time
    v_final = math.sqrt(vx ** 2 + vy_final ** 2)
    print(f"(c) 충돌 순간 속도 크기: {v_final:.2f} m/s")

    # 7 충돌 순간 각도 계산 (지면과 이루는 각)
    angle_impact_rad = math.atan2(vy_final, vx)
    angle_impact_deg = math.degrees(angle_impact_rad)
    print(f"(d) 충돌 순간 각도: {angle_impact_deg:.2f} 도")

if __name__ == "__main__":
    main()
```
